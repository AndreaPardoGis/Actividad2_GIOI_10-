# 📊 Análisis y visualización de dióxido de nitrógeno (NO₂) en 2019
Este notebook corresponde a la **Fase 2 del trabajo en grupo**, centrado en la visualización de los datos limpios de contaminación en **Oxford Street (Westminster), Londres** durante el año 2019.

Se utilizan los datos ya procesados por el compañero encargado de la limpieza, en formato `.xlsx`.

### Objetivo de esta fase:
- Graficar la evolución del dióxido de nitrógeno (NO₂) a lo largo del año.
- Identificar y marcar el valor mínimo, máximo, promedio (media) y mediana.


## 📦 Librerías utilizadas
Usamos las siguientes librerías de Python para el análisis y visualización:

- `pandas`: para manipulación de datos.
- `matplotlib` y `seaborn`: para crear los gráficos.
- `openpyxl`: para poder leer archivos `.xlsx`.

Estas librerías permiten una visualización clara y profesional de los datos, facilitando la interpretación.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Carga de archivos en excel
df = pd.reas_excel("archivo.xlsx", engine="openpyxl")
```

## 📥 Carga y revisión de los datos
Cargamos el archivo Excel con los datos limpios. La tabla debe incluir:

- Una columna con la **fecha y hora** (por ejemplo: `datetime`)
- Una columna con los valores horarios de **NO₂ (μg/m³)**

Aseguramos que los datos estén ordenados cronológicamente antes de visualizar.


```python
# Ajustamos los nombres de las columnas, si es necesario, para que coincidan con los de nuestro archivo
df.rename(columns={
    df.columns[0]: "columna1", #primera columna como datatime
    df.columns[1]: "columna2", #segunda columna como NO2
}, inplace=True)
```

```python
# ordenamos cronologicamente, si es necesario
df = df.sort_values(by="columna1")
```

```python
# Estadísticas
min_value = df["NO2"].min()
max_value = df["NO2"].max()
mean_value = df["NO2"].mean()
median_value = df["NO2"].median()
min_date = df[df["NO2"] == min_value]["datetime"].values[0]
max_date = df[df["NO2"] == max_value]["datetime"].values[0]
```

```python
# Gráfico de línea
plt.figure(figsize=(14,6))
sns.lineplot(data=df, x="datetime", y="NO2", label="NO₂ (μg/m³)", linewidth=1.5)
```

```python
# Líneas y puntos importantes
plt.axhline(mean_value, color='orange', linestyle='--', label=f"Media: {mean_value:.2f}")
plt.axhline(median_value, color='purple', linestyle='--', label=f"Mediana: {median_value:.2f}")
plt.scatter(min_date, min_value, color='blue', label=f"Mínimo: {min_value:.2f}", zorder=5)
plt.scatter(max_date, max_value, color='red', label=f"Máximo: {max_value:.2f}", zorder=5)
```

## 📈 Gráfico de línea con NO₂ durante el año 2019

En este gráfico mostramos la evolución de los valores horarios de NO₂.

También se indican:

- **Línea discontinua naranja**: media anual
- **Línea discontinua morada**: mediana
- **Punto azul**: valor mínimo
- **Punto rojo**: valor máximo

Esto nos permite identificar visualmente los momentos más y menos contaminantes del año.


```python
# Estética
plt.title("Concentración de Dióxido de Nitrógeno (NO₂) en 2019 - Oxford Street, Westminster")
plt.xlabel("Fecha")
plt.ylabel("NO₂ (μg/m³)")
plt.legend()
plt.grid(True)
plt.tight_layout()

plt.show()
```

## ✅ Conclusión del análisis gráfico

Este gráfico permite observar cómo varió la contaminación por NO₂ a lo largo del año.

Los valores extremos nos ayudan a identificar **posibles episodios de alta o baja polución**, mientras que los valores medios nos permiten analizar la **tendencia general**. Esta información será útil para comparar con otras métricas como el día más contaminado o la distribución mensual en las siguientes tareas.
